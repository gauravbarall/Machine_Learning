```python
import pandas as pd
```

```python
import numpy as np

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression

import seaborn as sns
```

```python
df = pd.read_csv('train.csv')[['Age','Pclass','SibSp','Parch','Survived']]
```

```python
df.head()
```

```python
#check for missing values
df.isna().sum()
```

```python
# drop missing values
df.dropna(inplace = True)
```

```python
df.head()
```

```python
# splitting input and output features
X = df.iloc[:, :4]
y= df.iloc[:, -1:]
```

## Accuracy before Feature Construction

```python
np.mean(cross_val_score(LogisticRegression(), X, y.values.ravel(), scoring='accuracy', cv=20))
```

## Applying Feature Construction

```python
df['Family_size'] = df['SibSp'] + df['Parch'] + 1
```

```python
df.head()
```

```python
def func(n):
    if n == 1:
        # alone
        return 0
    elif n > 1 and n <= 4:
        #small family
        return 1
    else:
        #large family
        return 2
```

```python
df['Family_type'] = df['Family_size'].apply(func)
```

```python
df.head()
```

```python
df.drop(columns=['SibSp', 'Parch', 'Family_size'], inplace=True)
```

```python
df.head()
```

```python
# splitting input and output features
X = df.drop(columns = ['Survived'])
```

```python
y = df.iloc[:, 2]
```

## Accuracy after feature construction

```python
np.mean(cross_val_score(LogisticRegression(), X, y.values.ravel(), scoring='accuracy', cv=20))
```
